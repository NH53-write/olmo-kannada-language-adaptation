```python
import json
import pandas as pd
from tabulate import tabulate
```

```python
with open("/workspace/language_adaptation/eval/bpb_tokenizer_variants.json", encoding="utf-8") as f:
    results = json.load(f)
```

```python
df = pd.DataFrame([
    {
        "Tokenizer": name,
        "New tokens": m["new_tokens"],
        "BPB": m["bits_per_byte"],
        "Bits/char": m["bits_per_char"],
        "Chars/token": m["chars_per_token"],
        "Tokens (eval set)": m["tokens"],
    }
    for name, m in results.items()
]).set_index("Tokenizer")
```

```python
base_bpb = df.loc["base", "BPB"]
df["BPB vs base"] = df["BPB"] - base_bpb
```

```python
print(tabulate(df.round(4), headers="keys", tablefmt="grid"))
```
